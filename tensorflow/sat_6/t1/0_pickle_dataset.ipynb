```python
import os
import pickle
import numpy as np
import pandas as pd

from utils import load_sat6_csv, pad_to_32x32
```

```python
DATA_DIR = "./dataset"

TRAIN_PICKLE_PATH = "./dataset/sat6_train.pkl"
TEST_PICKLE_PATH = "./dataset/sat6_test.pkl"
```

```python
X_train, y_train = load_sat6_csv(
    os.path.join(DATA_DIR, "X_train_sat6.csv"),
    os.path.join(DATA_DIR, "y_train_sat6.csv"),
)
X_test, y_test = load_sat6_csv(
    os.path.join(DATA_DIR, "X_test_sat6.csv"),
    os.path.join(DATA_DIR, "y_test_sat6.csv"),
)

print("Formato treino (original 28x28):", X_train.shape, y_train.shape)
print("Formato teste  (original 28x28):", X_test.shape, y_test.shape)

X_train = pad_to_32x32(X_train)
X_test = pad_to_32x32(X_test)

print("Formato treino (após padding 32x32):", X_train.shape)
print("Formato teste  (após padding 32x32):", X_test.shape)

with open(TRAIN_PICKLE_PATH, "wb") as f:
    pickle.dump({"X": X_train, "y": y_train}, f, protocol=pickle.HIGHEST_PROTOCOL)

with open(TEST_PICKLE_PATH, "wb") as f:
    pickle.dump({"X": X_test, "y": y_test}, f, protocol=pickle.HIGHEST_PROTOCOL)
```
